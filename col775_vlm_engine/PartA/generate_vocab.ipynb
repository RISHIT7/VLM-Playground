# Generate Vocabulary for Autograder

This notebook builds the vocabulary from your training captions and saves it as `vocab.json`. This file is required by the `inference_retrieval.py` script for CLIP evaluation.

```python
import os
import sys
import json

# Ensure the PartA directory is in the path so we can import our local data modules
sys.path.append(os.path.abspath('.'))

from data.text_tokenizer import CLEVRTokenizer
```

## 1. Set paths
Update `TRAIN_CAPTIONS_PATH` to point to your training JSON file (e.g., `clevr_train_captions.json`).

```python
TRAIN_CAPTIONS_PATH = '../data/train/clevr_train_captions.json' # Update this path
OUTPUT_VOCAB_PATH = './dummy_weights/vocab.json'               # Output path for autograder folder

if not os.path.exists(TRAIN_CAPTIONS_PATH):
    print(f"ERROR: Training captions not found at {TRAIN_CAPTIONS_PATH}")
else:
    print("Found training captions file.")
```

## 2. Build and Save Vocabulary

```python
tokenizer = CLEVRTokenizer(max_seq_len=77)

print("Building vocabulary...")
tokenizer.build_vocab(TRAIN_CAPTIONS_PATH)

print(f"Vocabulary size: {tokenizer.vocab_size}")

# Use the new save_vocab helper I added to your tokenizer script
os.makedirs(os.path.dirname(OUTPUT_VOCAB_PATH), exist_ok=True)
tokenizer.save_vocab(OUTPUT_VOCAB_PATH)

print(f"Successfully saved vocab to {OUTPUT_VOCAB_PATH}")
```

## 3. Sanity Check
Verify the saved JSON is correct.

```python
with open(OUTPUT_VOCAB_PATH, 'r') as f:
    vocab = json.load(f)

print("Sample vocab entries:")
items = list(vocab.items())
for word, idx in items[:10]:
    print(f"  {word}: {idx}")
```
